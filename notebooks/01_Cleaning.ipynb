# Data Cleaning and Feature Preparation

This notebook loads the raw dataset generated by the scraping script and performs data cleaning,
type standardization, and basic feature engineering to prepare the dataset for analysis,
A/B testing simulation, and a baseline machine learning model.

Input:
- `data/raw/books_raw.csv`

Output:
- `data/processed/books_clean.csv`


```python
import pandas as pd
import numpy as np
from pathlib import Path

# Project root (notebook is inside /notebooks)
PROJECT_ROOT = Path("..").resolve()

raw_path = PROJECT_ROOT / "data" / "raw" / "books_raw.csv"
df = pd.read_csv(raw_path)

df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
rating_map = {"One": 1, "Two": 2, "Three": 3, "Four": 4, "Five": 5}
df["rating_num"] = df["rating"].map(rating_map)

df[["rating", "rating_num"]].head()
```

```python
df["title_length"] = df["title"].str.len()
df["high_rating"] = (df["rating_num"] >= 4).astype(int)
```

Price values were already standardized during the extraction step. Pandas correctly inferred the column as float64 when loading the dataset, so no additional type casting was required during cleaning.

Rating values were successfully mapped from categorical labels to numeric values. No missing values were introduced during the transformation.

```python
processed_dir = PROJECT_ROOT / "data" / "processed"
processed_dir.mkdir(parents=True, exist_ok=True)

clean_path = processed_dir / "books_clean.csv"
df.to_csv(clean_path, index=False)

clean_path
```
