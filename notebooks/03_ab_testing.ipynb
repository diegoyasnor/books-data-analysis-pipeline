# Simulated A/B Testing

This notebook demonstrates a simulated A/B testing framework to evaluate whether
highlighting highly-rated books could improve conversion rates.

Since real user-level conversion data is not available, conversion outcomes are
simulated to illustrate experimental design, metric definition, and statistical
decision-making.


```python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.stats import ttest_ind

PROJECT_ROOT = Path("..").resolve()
df = pd.read_csv(PROJECT_ROOT / "data" / "processed" / "books_clean.csv")

df.head()
```

## Hypothesis

**Null hypothesis (H0):**  
Displaying a "Highly Rated" badge does not change the conversion rate.

**Alternative hypothesis (H1):**  
Displaying a "Highly Rated" badge increases the conversion rate.


```python
np.random.seed(42)
df["group"] = np.random.choice(["A","B"], size = len(df))
df["group"].value_counts()
```

I simulated a scenario where variant B had a 1 percentage point higher conversion rate to validate the A/B testing framework.

```python
control_rate= 0.05
variant_rate= 0.06

df["conversion"] = np.where(df["group"] =="A",
                   np.random.binomial(1,control_rate,size =len(df)),
                   np.random.binomial(1,variant_rate, size = len(df)) 
                           )
df[["group","conversion"]].head()
```

```python
control = df[df["group"] == "A"]["conversion"]
variant = df[df["group"] == "B"]["conversion"]

t_stat, p_value = ttest_ind(variant, control, equal_var=False)
t_stat, p_value
```

## Statistical Decision

The p-value obtained from the test is higher than the 5% significance threshold.
Therefore, we fail to reject the null hypothesis.

This suggests that the observed difference in conversion rates between the
control and variant groups could be due to random variation rather than a
true effect of the treatment.


## Conclusion

In this simulated experiment, highlighting highly-rated books did not lead
to a statistically significant increase in conversion rates.

From a business perspective, this result suggests that additional testing,
larger sample sizes, or alternative treatments may be required before
making a product decision.
