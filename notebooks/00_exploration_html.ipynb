# HTML Exploration and Selector Validation

This notebook was used exclusively for **initial exploration of the website's HTML structure** 
(Books to Scrape) in order to identify and validate the CSS selectors required for data extraction.

The goal of this exploration was to:
- Understand the repeated product card structure
- Locate key fields such as title, price, rating, and product links
- Validate selectors on a single page before scaling the extraction

⚠️ **Important note**  
This notebook is **not** the final scraping implementation.

The production-ready and reproducible data extraction logic is implemented as a standalone Python script:

`src/extract/scrape_all_pages.py`


```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
```

```python
url= "https://books.toscrape.com/catalogue/page-1.html"
html= requests.get(url,timeout=30).text
soup= BeautifulSoup(html, "html.parser")
```

```python
books= []
for card in soup.select("article.product_pod"):
    title = card.select_one("h3 a").get("title")
    price  = card.select_one("p.price_color").get_text(strip =True).replace("Â", "").replace("£", "")
    
    rating_tag = card.select_one("p.star-rating")
    classes = rating_tag.get("class",[])
    rating = next((c for c in classes if c != "star-rating"), None)

    
    href = card.select_one("h3 a").get("href")
    link= urljoin(url,href)
    books.append({"title": title,  "price_gbp": price, "rating":rating, "url": link})

print ("books:", len(books))
print(books[0])
```
